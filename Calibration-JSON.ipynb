## Calibration

* In this scripot we define the used small molecules (substrate, buffer etc) and the protein used throughout our experiments. These are saved as json files.

* The script is structured as follwed: 
    1) Define used proteins (Uniprot ID)
    2) Define constant molecules (Puchem ID), these are molecules which do not change concentration through out the experiment, such as the buffer
    3) Define non-constant molecules (Pubchem ID), these are molecules which changes in concentration, and thus has a calibration curve (e.g. substrate(s) and product(s))

```python
from chromhandler import Molecule, Protein, ChromAnalyzer
from pathlib import Path
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from brokenaxes import brokenaxes

calib_path = Path("20260610_Standards")

wavelength = 254
 

target_RT = { #20260610_Standards
   "cGAMP": 2.596,
    "ADP" : 2.732,
    "GDP" : 3.721,
    "ATP" : 4.848,
    "GTP" : 6.694,
}

tolerances = {
   "cGAMP": 0.2,
    "ADP" : 0.2,
    "GDP" : 0.2,
    "ATP" : 0.2,  
    "GTP" : 0.4,
}

PUBCHEM_CIDS = {
    "cGAMP": 135564529,
    "ADP"  : 6022,
    "GDP"  : 135398619,
    "ATP"  : 5957,
    "GTP"  : 135398635,
    
}



for molecule_path in calib_path.iterdir():
    if not molecule_path.is_dir():
        continue
    for txt_file in sorted(molecule_path.glob("*.txt")):
        with open(txt_file, "r", encoding="utf-8") as f:
            lines = f.readlines()
        is_rdl = bool(lines) and lines[0].startswith("┌─────")
        has_sample_name = any("Sample Name" in line for line in lines)
        has_signal = any("Signal:" in line for line in lines)
        if not (is_rdl and has_sample_name and has_signal):
            print(f"PROBLEM: {txt_file}")
            print(f"  starts with box header: {is_rdl}")
            print(f"  has 'Sample Name' line: {has_sample_name}")
            print(f"  has 'Signal:' line:     {has_signal}")

    name = molecule_path.name

    if name not in PUBCHEM_CIDS:
        raise ValueError(f"No PubChem CID defined for '{name}'")

    analyzer = ChromAnalyzer.read_agilent(
        path=str(molecule_path),
        ph=7,
        temperature=37,
        mode="calibration",
    )

    molecule = analyzer.define_molecule(
        pubchem_cid=PUBCHEM_CIDS[name],
        id=name,
        name=name,  # PREVENTS PUBCHEM TIMEOUT
        retention_time=target_RT[name],
        retention_tolerance=tolerances[name],
        wavelength=wavelength,
    )

    analyzer.assign_all_peaks()

    analyzer.add_standard(
        molecule=molecule,
        wavelength=wavelength,
        visualize=True,
    )

    molecule.save_json(f"Results/Molecules/{molecule.id}.json")
```
