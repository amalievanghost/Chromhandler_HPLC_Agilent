## Sort the data files into subfolders

time_sample/
* _1/
* _2/
* _3/


Negative controls will stay in raw-data folder


First time setting up the workflow, you need to do the following steps:
1. Create a project environment in the terminal
2. Create the subfolder "Data", "Raw-data", and "Standards"
3. Sort the standards, for calibration


Step 2: Sort data into subfolders

```python
import shutil
import re
from pathlib import Path

# --- SETTINGS ---
SOURCE_PATH = Path("Raw-data/20251117_100mM-HEPES_0.02-tmmcGAS")  # CHANGE experiment
TARGET_PATH = Path("Data/time_sample")  # directly include the time_sample folder
MODE = "copy"  # or "move"

# --- REGEX ---
prefix_pattern = re.compile(r'^(?P<num>[1-3])_')

def sort_by_prefix(source_path: Path, target_path: Path, mode=MODE):
    print(f"Processing source folder: {source_path}")

    if not source_path.exists():
        print("ERROR: Source folder does not exist!")
        return

    source_folder_name = source_path.name

    # Create experiment root folder inside time_sample
    export_root = target_path / source_folder_name
    export_root.mkdir(parents=True, exist_ok=True)

    # Create subfolders only for 1, 2, 3
    subfolders = {}
    for name in ("1", "2", "3"):
        folder = export_root / f"{source_folder_name}_{name}"
        folder.mkdir(exist_ok=True)
        subfolders[name] = folder

    # Process files
    for file in source_path.iterdir():
        if not file.is_file():
            continue
        if file.suffix.lower() not in {".pdf", ".txt"}:
            continue

        m = prefix_pattern.match(file.name)
        if not m:
            print(f"Skipping (kept in main folder): {file.name}")
            continue

        folder_key = m.group("num")
        dst = subfolders[folder_key] / file.name

        try:
            if mode == "copy":
                shutil.copyfile(file, dst)  # simpler copy (no metadata)
            else:
                shutil.move(str(file), str(dst))
            print(f"{mode.capitalize()}: {file.name} → {folder_key}")
        except PermissionError:
            print(f"⚠ Permission denied while copying {file.name}, skipping.")

    print("✔ Sorting complete")


if __name__ == "__main__":
    sort_by_prefix(SOURCE_PATH, TARGET_PATH, MODE)
```
